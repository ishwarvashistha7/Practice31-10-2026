```python
#QUES6:-Load a JSON file with json.loads() and convert it to a DataFrame
#Then report the DatFrame back to JSON usin to_json(orient="records").
```

```python
#QUES7:-Read a large CSV in chunks of 1000 rows(chunksize) and find the average of one column across all chunks.
```

```python
chunk_size = 1000
chunks = pd.read_csv('students.csv', chunksize=chunk_size)
total_sum = 0
total_count = 0
for chunk in chunks:
    total_sum += chunk['Marks'].sum()
    total_count += len(chunk)
overall_average = total_sum / total_count
print(f"Overall Average of Marks: {overall_average}")
```

```python
#Read two sheets from an Excel file using pd.ExcelFile and pd.read_excel().Write a DataFrame to Excel with to_excel().
```

```python
with pd.ExcelWriter('students_data.xlsx', engine='openpyxl') as writer:
    df_original.to_excel(writer, sheet_name='Sheet1', index=False)
    df_original[['Name', 'Marks']].to_excel(writer, sheet_name='Sheet2', index=False)
excel_file = pd.ExcelFile('students_data.xlsx')
print("Available Sheets:", excel_file.sheet_names)
sheet1_df = pd.read_excel(excel_file, sheet_name='Sheet1')
sheet2_df = pd.read_excel(excel_file, sheet_name='Sheet2')
print("\n--- Sheet 1 Data ---")
print(sheet1_df.head(3))
print("\n--- Sheet 2 Data ---")
print(sheet2_df.head(3))
```

```python
#Ques9:-Use pd.read_html() on a Wikipedia page containing a table .Find how many tables were returned and dispay the one you need.
```

```python
import urllib.request
url = 'https://en.wikipedia.org/wiki/List_of_Asian_countries_by_area'
req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
html_content = urllib.request.urlopen(req).read()
tables = pd.read_html(html_content)
print(f"Total tables found: {len(tables)}")
required_table = tables[0]
print("\n--- Displaying Required Table ---")
print(required_table.head())
```

```python
#QUES10:-Save a DataFrame with to_pickle() and load it back woth read_pickle() . Compare the filesizes of the CSVand pickle versions.
```

```python
import os
df_original.to_pickle('students.pkl')
df_loaded = pd.read_pickle('students.pkl')
csv_size = os.path.getsize('students.csv')
pkl_size = os.path.getsize('students.pkl')
print("--- Data Loaded from Pickle ---")
print(df_loaded.head(3))
print("\n--- File Size Comparison ---")
print(f"CSV File Size    : {csv_size} bytes")
print(f"Pickle File Size : {pkl_size} bytes")
```

```python
#QUES11:-Save a DataFrame in Parquet or HDF5 format and read it back State one advantage over CSV.
```

```python
df_original.to_parquet('students.parquet')
df_parquet = pd.read_parquet('students.parquet')
print("--- Data Loaded from Parquet ---")
print(df_parquet.head(3))
```
